```python
#Linear Search:
def linear(arr,target):
    for i in range(len(arr)):
        if arr[i]==target:
            return i
    return -1
arr=[1,2,3,4,5,6]
print(linear(arr,4))    
```

```python
#Binary Search:
def binary(arr,key):
    low=0
    n=len(arr)
    high=n-1

    while low<=high:
        mid=(low+high)//2
        if arr[mid]==key:
            return mid
        elif arr[mid]<key:
            low==mid+1
        else:
            high=mid-1
    return -1

arr=[1,2,3,4,5,6,7]
print(binary(arr,4))        
```

```python
#Recursion:
def factorial(n):
    if n == 0 or n == 1:   # Base case
        return 1
    return n * factorial(n - 1)

# Example
print(factorial(5))
```

```python
#Prefix:
def prefix_sum(arr):
    prefix = [0] * len(arr)
    prefix[0] = arr[0]

    for i in range(1, len(arr)):
        prefix[i] = prefix[i - 1] + arr[i]

    return prefix

# Range sum query
def range_sum(prefix, left, right):
    if left == 0:
        return prefix[right]
    return prefix[right] - prefix[left - 1]

# Example
arr = [2, 4, 6, 8]
prefix = prefix_sum(arr)
print(prefix)
print(range_sum(prefix, 1, 3))
```

```python
#Sliding window :
def max_sub(arr,k):
    window_sum=sum(arr[:k])
    max_sum=window_sum

    for i in range(k,len(arr)):
        window_sum += arr[i]
        window_sum -=arr[i-k]
        max_sum =max(max_sum,window_sum)
    return max_sum

arr=[1,2,3,4,5,6,7,8]
print(max_sub(arr,3))    
```
